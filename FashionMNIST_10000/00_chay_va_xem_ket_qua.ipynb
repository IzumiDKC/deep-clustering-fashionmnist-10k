```python
import pandas as pd
from pathlib import Path
folder=Path("results")
runs=pd.read_csv(folder/"runs.csv")
summary=pd.read_csv(folder/"summary.csv")
display(runs[["method","seed","ACC","NMI","ARI","train_seconds"]])
display(summary)
```

```python
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,3,figsize=(14,4))
for ax,col in zip(axes,["acc","nmi","ari"]):
    x=summary[col+"_mean"].values
    e=summary[col+"_std"].fillna(0).values
    ax.barh(summary["method"],x,xerr=e,capsize=3)
    ax.set_title(col.upper()+" (mean ± SD, 3 seed)")
    ax.set_xlim(0,100)
plt.tight_layout();plt.show()
```
